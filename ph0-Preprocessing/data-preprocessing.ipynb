```python
import json
import random
from pathlib import Path


def create_mini_datasets(datasets, output_dir="/kaggle/working/data/raw", seed=42):

    output_dir = Path(output_dir)
    output_dir.mkdir(parents=True, exist_ok=True)

    rng = random.Random(seed)

    for name, config in datasets.items():

        input_path = Path(config["path"])
        n_samples = config["n_samples"]

        print(f"\nProcessing: {name}")

        with input_path.open("r", encoding="utf-8") as f:
            data = [json.loads(line) for line in f]

        print(f"  Original: {len(data):,}")

        if n_samples > len(data):
            raise ValueError(
                f"{name}: requested {n_samples:,} "
                f"but dataset only contains {len(data):,}"
            )

        sampled = rng.sample(data, n_samples)

        output_path = output_dir / f"{name}_mini.jsonl"
        with output_path.open("w", encoding="utf-8") as f:
            for example in sampled:
                f.write(
                    json.dumps(
                        example,
                        ensure_ascii=False
                    ) + "\n"
                )

        print(f"  Selected: {len(sampled):,}")
        print(f"  Saved:    {output_path}")
        

datasets = {

    "flan_v2": {
        "path": "/kaggle/input/datasets/tanmairaghava/flan-v2/flan_v2_data.jsonl",
        "n_samples": 3000,
    },

    "cot": {
        "path": "/kaggle/input/datasets/tanmairaghava/cot-data/cot_data.jsonl",
        "n_samples": 3000,
    },

    "dolly": {
        "path": "/kaggle/input/datasets/tanmairaghava/dolly-data/dolly_data.jsonl",
        "n_samples": 1500,
    },

    "oasst1": {
        "path": "/kaggle/input/datasets/tanmairaghava/oasst-data/oasst1_data.jsonl",
        "n_samples": 1500,
    },
}


create_mini_datasets(datasets, output_dir="/kaggle/working/data/raw", seed=42)
```
